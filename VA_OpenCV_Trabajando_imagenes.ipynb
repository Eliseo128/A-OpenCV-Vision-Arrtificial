```python
import cv2
```

```python
img = cv2.imread('/content/sample_data/girl-original.jpeg')
```

```python
img
```

Lets look at what we are working with and the shape!

```python
print(type(img))
print(img.shape)
```

```python
from google.colab.patches import cv2_imshow
cv2_imshow(img)
```

```python
small = cv2.resize(img, (0,0), fx=0.5, fy=0.5)
```

```python
from google.colab.patches import cv2_imshow
cv2_imshow(small)
```

```python
small.shape
```

Take a look at the first pixel! OpenCV2 - BGR

```python
small[0,0]
```

Lets give it a new value

```python
small[0,0] = (255,255,255)
```

```python
small[0,0]
```

```python
from google.colab.patches import cv2_imshow
cv2_imshow(small)
```

Lets change the first 100 rows and 10 colums to red

```python
small[0:100,0:10] = (0,0,255)
```

```python
from google.colab.patches import cv2_imshow
cv2_imshow(small)
```

Grayscale Image

```python
gray = cv2.cvtColor(small, cv2.COLOR_BGR2GRAY)
gray.shape
```

```python
from google.colab.patches import cv2_imshow
cv2_imshow(gray)
```

Save Image

```python
cv2.imwrite("Gray.png", gray)
```
