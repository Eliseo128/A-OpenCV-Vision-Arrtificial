### Lines on the image


```python
I = cv2.imread('JohnLennon.jpg')
plt.imshow(I)
x = [100,100,500,500,100]
y = [500,100,100,500,500]
plt.plot(x, y, c='blue')
plt.show()
```

```python
I = cv2.imread('JohnLennon.jpg')
plt.imshow(I)

N = 30 # sides of the polygon
x = np.zeros((N+1,))
y = np.zeros((N+1,))
(xc,yc) = (340,320)
r = 150 # radius
for i in range(N+1):
  t    = 2*np.pi/N*i
  x[i] = xc + r*np.cos(t)
  y[i] = yc + r*np.sin(t)

plt.plot(x, y, c='orange')
plt.show()
```
